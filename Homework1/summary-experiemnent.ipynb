# Summary of Experiment — Gemma 3 (4B)

## 1. Objective
Fine-tune Gemma 3 on two Khmer Q&A datasets, compare training and validation loss, and test the quality of generated answers.

## 2. Experiment setup

| Item | Banking | General knowledge |
|---|---|---|
| Dataset | `data_kh.csv` | `data_kh_general.csv` |
| Total records | 519 | 519 |
| Training records | 415 | 415 |
| Validation records | 104 | 104 |
| Training steps | 60 | 60 |
| Saved adapter | `gemma-3` | `gemma-3-general` |

Both experiments use Gemma 3 (4B). The comparison uses a fixed split seed (3407), and both saved models are loaded in 4-bit mode for inference.

## 3. Loss comparison

| Step | Banking training | Banking validation | General training | General validation |
|---:|---:|---:|---:|---:|
| 10 | 1.810252 | 2.000398 | 1.526133 | 1.675640 |
| 20 | 1.569814 | 1.669297 | 1.555844 | 1.493430 |
| 30 | 1.212971 | 1.526837 | 1.393564 | 1.430233 |
| 40 | 1.496915 | 1.460079 | 1.454180 | 1.393333 |
| 50 | 1.516805 | 1.425591 | 1.501650 | 1.376544 |
| 60 | 1.411848 | 1.416403 | 1.415534 | 1.370518 |

- **Banking:** Training loss generally decreases, with some fluctuations. Validation loss decreases at every checkpoint.
- **General:** Training loss fluctuates slightly, while validation loss decreases steadily.
- **Best validation step:** 60 for both models.
- **Final result:** Banking has slightly lower training loss, while General has lower validation loss.

## 4. Overfitting and underfitting

- **Overfitting:** No clear sign in either experiment because validation loss keeps decreasing at the recorded checkpoints.
- **Underfitting:** Both models show learning, but the loss curves alone cannot rule out underfitting. Validation loss is still improving at step 60, so additional training could be tested.
- The lower validation loss of the General model does not prove it is better overall because the validation datasets are different.

## 5. Response comparison

Both models answered the same 20 held-out questions: 10 banking and 10 general knowledge. Inference used greedy decoding and a maximum of 256 new tokens.

- **Banking model:** Gives some relevant banking answers but also makes factual mistakes. For example, it claims a membership card has an annual fee of 50,000 riel, while the reference says it is free.
- **General model:** Produces relevant answers on some general topics, but also makes mistakes on banking questions, including repetitive text in the membership-card answer.
- **Both models:** Incorrectly state that the National Bank of Cambodia provides loans to the public, contrary to the reference answer.

The response table was exported to `model_comparison_responses.csv`. Correctness, relevance, completeness and Khmer quality have not yet been scored. Therefore, no overall accuracy or response-quality winner can be claimed.

## 6. Conclusion

Both models were fine-tuned and tested successfully. Training and validation losses improved overall, with no clear overfitting during the 60 steps. The General model achieved lower final validation loss (1.370518) than the Banking model (1.416403), but the datasets differ in content and difficulty.

The response examples show that lower loss does not guarantee correct answers. Manual evaluation is needed to determine which model performs better in each domain.

**Limitation:** Only six common loss checkpoints are available, and the displayed response previews are truncated. A longer training experiment and completed response scoring would provide stronger evidence.
